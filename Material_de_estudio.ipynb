# Capítulo 1: Números, expresiones y strings
## Operadores aritméticos

| Operador | Qué hace | Ejemplo |
|---|---|---|
| `+` | Suma | `30 + 12` |
| `-` | Resta | `43 - 1` |
| `*` | Multiplicación | `6 * 7` |
| `/` | División (**siempre da float**) | `84 / 2` → `42.0` |
| `//` | División entera (redondea hacia abajo) | `85 // 2` → `42` |
| `**` | Exponenciación | `7 ** 2` → `49` |
| `^` | **NO es exponenciación** — es XOR (bitwise). No lo usaremos, pero recuerda que existe. |


````python
print(30 + 12)
print(43 - 1)
print(6 * 7)
print(84 / 2)    # float, aunque el resultado sea exacto
print(85 // 2)   # entero, redondea hacia abajo
print(7 ** 2)
````

## Orden de operaciones

Python sigue el mismo orden que en matemática: exponenciación → multiplicación/división → suma/resta. Los paréntesis siempre van primero.

````python
print(12 + 5 * 6)      # multiplicación primero: 12 + 30 = 42
print((12 + 5) * 6)    # paréntesis primero: 17 * 6 = 102
print(6 + 6 ** 2)      # exponente primero: 6 + 36 = 42
````

## Funciones aritméticas básicas

* `round(x)` — redondea al entero más cercano. **Ojo con los .5**: redondea al par más cercano (no siempre hacia arriba).
* `abs(x)` — valor absoluto (siempre positivo o cero).

Toda llamada a función necesita paréntesis, incluso sin argumentos entre ellos: `abs 42` (sin paréntesis) es un `SyntaxError`.

````python
print(round(42.4))
print(round(42.6))
print(round(42.5))   # 42, porque 42 es el par más cercano
print(round(43.5))   # 44, porque 44 es el par más cercano
print(abs(-42))
print(abs(42))
````

````python
%%expect SyntaxError

abs 42   # falta el paréntesis
````

## Tipos y conversión

Los tres tipos vistos hasta ahora: `int`, `float`, `str`. Usa `type()` para consultar el tipo de cualquier valor.

`int()`, `float()` y `str()` convierten entre tipos. **Ojo**: un string con dígitos (`'126'`) no es un número — hay que convertirlo antes de operar.

````python
print(type(2))
print(type(42.0))
print(type('126'))     # str, aunque parezca un número
print(int(42.9))       # trunca, no redondea: 42
print(float(42))
print(int('126') / 3)  # convertir antes de operar
````

# Capítulo 2: Variables y sentencias
## Variables y asignación

Una **sentencia de asignación** crea una variable y le da un valor: `nombre = expresión`. No muestra nada en pantalla, pero tiene el efecto de crear la variable.

````python
n = 17
pi = 3.141592653589793
message = 'And now for something completely different'

print(n + 25)
print(2 * pi)
print(len(message))
````

## Nombres de variables

* Pueden tener letras, números y `_`, pero **no pueden empezar con número**.
* No pueden ser una **keyword** (`class`, `for`, `import`, etc — hay 35 en total).
* No pueden tener otra puntuación (`!`, espacios, etc).

````python
%%expect SyntaxError

million! = 1000000   # ! no es válido
````

````python
%%expect SyntaxError

76trombones = 'big parade'   # no puede empezar con número
````

````python
%%expect SyntaxError

class = 'Self-Defence Against Fresh Fruit'   # class es keyword
````

## El import statement y los módulos

Un **módulo** es una colección de variables y funciones. Para usarlas hay que importar el módulo, y acceder con el **operador punto** (`.`).

````python
import math

print(math.pi)
print(math.sqrt(25))
print(math.pow(5, 2))   # math.pow SIEMPRE devuelve float
````

## Expresiones vs. sentencias

* **Expresión**: tiene un valor (`n + 25`, `math.sqrt(25)`).
* **Sentencia**: tiene un efecto, no necesariamente un valor (`n = 17`, `import math`).

**Evaluar** una expresión = calcular su valor. **Ejecutar** una sentencia = correrla y que produzca su efecto.

## La función print

Si evalúas varias expresiones en una celda, **solo se muestra la última**. Para mostrar varios valores, usa `print()`.

````python
n + 2
n + 3    # solo se muestra este resultado: 20
````

````python
print(n + 2)   # ambos se muestran
print(n + 3)
````

````python
print('El valor de pi es', math.pi)   # print pone un espacio entre los valores
````

## Debugging: los 3 tipos de error

| Tipo | Cuándo ocurre | Ejemplo |
|---|---|---|
| **SyntaxError** | El programa ni siquiera puede empezar a correr; la estructura del código es inválida. | `million! = 1000000` |
| **Error de ejecución (runtime error / exception)** | El programa corre pero se detiene por un problema durante la ejecución. Incluye `TypeError`, `NameError`, etc. | `'126' / 3` |
| **Error semántico** | El programa corre completo, sin ningún mensaje de error, pero **no hace lo que se esperaba**. | `1 + 3 / 2` para calcular el promedio de 1 y 3 (da 2.5, no 2) |


````python
# Ejemplo de error semántico: no hay ningún mensaje de error, pero el resultado es incorrecto
promedio_mal = 1 + 3 / 2
promedio_bien = (1 + 3) / 2
print(promedio_mal)   # 2.5, incorrecto
print(promedio_bien)  # 2.0, correcto
````

# Capítulo 3: Funciones
## Definir una función

* `def` empieza la definición. El **header** (primera línea) termina en `:`.
* El **body** (cuerpo) va indentado (4 espacios por convención).
* Definir una función crea un **function object**; llamarla ejecuta su cuerpo.

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")

print_lyrics()   # la llamada ejecuta el cuerpo
````

## Parámetros y argumentos

El **parámetro** es el nombre usado dentro de la función. El **argumento** es el valor real que se entrega al llamarla.

````python
def print_twice(string):
    print(string)
    print(string)

print_twice('Ni!')          # 'Ni!' es el argumento; se asigna al parámetro 'string'
````

## Funciones que llaman a otras funciones

Una función puede usar otra dentro de su cuerpo. Esto evita repetir código.

````python
def repeat(word, n):
    print(word * n)

def first_two_lines():
    repeat('Spam, ', 4)
    repeat('Spam, ', 4)

first_two_lines()
````

## Repetición con for

`for variable in range(n):` repite el cuerpo `n` veces, con `variable` tomando los valores `0, 1, ..., n-1`.

````python
for i in range(4):
    print(i)
````

````python
def print_n_verses(n):
    for i in range(n):
        print_lyrics()
        print()

print_n_verses(2)
````

## Variables locales

Una variable creada **dentro** de una función es **local**: solo existe mientras la función se está ejecutando. Fuera de la función, usarla produce un `NameError`.

````python
def cat_twice(part1, part2):
    cat = part1 + part2   # cat es local: solo existe aquí adentro
    print_twice(cat)

cat_twice('Always look on the ', 'bright side of life.')
````

````python
%%expect NameError

print(cat)   # cat no existe fuera de cat_twice
````

## Stack diagrams (diagramas de pila)

Un **stack diagram** muestra qué función llamó a cuál. Cada función tiene un **frame**: una caja con sus parámetros y variables locales adentro. El frame de `__main__` (el nivel más externo) va en la base; cada función llamada se apila encima.

## Tracebacks

Cuando ocurre un error dentro de una función, Python muestra un **traceback**: la lista de funciones que se estaban ejecutando, desde la más antigua (arriba) hasta la que estaba corriendo cuando ocurrió el error (abajo).

````python
def print_twice(string):
    print(cat)    # error a propósito: cat no existe aquí
    print(cat)

def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

````python
%%expect NameError

cat_twice('Always look on the ', 'bright side of life.')
# El traceback mostrará: cat_twice llamó a print_twice, y el error ocurrió en print_twice
````

## ¿Por qué usar funciones?

* Nombrar un grupo de instrucciones hace el programa más fácil de leer.
* Evitan repetir código: si cambias algo, lo cambias en un solo lugar.
* Permiten dividir un programa grande en partes que se pueden depurar por separado.
* Una función bien escrita se puede reutilizar en otros programas.

## Debugging (capítulo 3)

* Si el error **muestra un mensaje** (SyntaxError, NameError, TypeError...), lee el traceback: te dice qué función estaba corriendo y por qué.
* Si el error **no muestra ningún mensaje** pero el resultado está mal, es un **error semántico**: agrega `print()` temporales dentro de la función para revisar los valores paso a paso.

|---|---|---|
| `SyntaxError` | La estructura del código es inválida: nombre ilegal, falta paréntesis, comillas inválidas, keyword como nombre, indentación faltante. | `million! = 5`, `abs 42`, `` `Hello` ``, `class = 5` |
| `TypeError` | Un operador o función recibe un valor del tipo incorrecto, o la cantidad de argumentos incorrecta. | `'126' / 3`, `math.pow(2)`, `math.sqrt('123')` |
| `NameError` | Se usa una variable que no existe (o ya no existe, por ser local de otra función). | usar `cat` fuera de la función donde se creó |
| Error semántico | El código corre sin ningún mensaje de error, pero no hace lo que se esperaba. | `1 + 3 / 2` para calcular un promedio |
