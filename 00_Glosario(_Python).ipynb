#Glosario

1. División entera (`//`)
* Es una división entre 2 numeros la cual su resultado siempre es entero, por lo tanto se redondea al entero menor.
* Ej: 41 // 4 = 10
---
2. Expresión
* Colección de operadores y números.
* Ej: 42 + 3 * 4
---
3. Valor
* Es el resultado de una expresión
---
4. Error: SyntaxError
+ Ocurre cuando llamo una función y cometo un error en su escritra
* Ej: abs 50, es un eror ya que no puse "(50)"
---
5. Función (`len`)
- Cuenta la cantidad de caracteres que tiene un string. Si pongo los parentesis tengo un error de sintaxis.
---
6. Error: TypeError
- Ocurre cuando se ejecuta una operación que no está asociada a un tipo
- Ej: dividir un string, pues con string solo puedo "*" y "+"
---
7. Debugging
- Proceso de rastrear y corregir errores (bugs)
